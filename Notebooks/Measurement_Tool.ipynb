```python
# @title
%matplotlib widget

!pip -q install ipympl

from google.colab import output
output.enable_custom_widget_manager()

import matplotlib.pyplot as plt
import numpy as np
from imageio.v3 import imread

# Load sample image
img = imread(FILENAME)

fig, ax = plt.subplots(figsize=(10,10))
ax.imshow(img)
ax.set_title("Click two points to measure distance")

coords = []
texts = [] # To store text annotations
lines = [] # To store line objects
point_markers = [] # To store point marker objects

def onclick(event):
    if event.xdata is not None and event.ydata is not None:
        # Clear previous annotations if starting a new measurement
        if len(coords) == 0:
            for text_obj in texts:
                text_obj.remove()
            for line_obj in lines:
                line_obj.remove()
            for marker_obj in point_markers:
                marker_obj.remove()
            texts.clear()
            lines.clear()
            point_markers.clear()
            # Reset title to original when starting a new measurement
            ax.set_title("Click two points to measure distance")
            fig.canvas.draw()

        coords.append((event.xdata, event.ydata))
        point_marker, = ax.plot(event.xdata, event.ydata, 'ro') # Draw red dot
        point_markers.append(point_marker)

        # Add text annotation for the point
        text = ax.text(event.xdata + 10, event.ydata + 10, f'({event.xdata:.0f}, {event.ydata:.0f})', color='white')
        texts.append(text)

        print(f"Point {len(coords)} (x,y): ({event.xdata:.2f}, {event.ydata:.2f})")

        if len(coords) == 2:
            pt1, pt2 = coords
            pixel_dist = np.sqrt((pt2[0] - pt1[0])**2 + (pt2[1] - pt1[1])**2)
            print(f"Distance: {pixel_dist:.2f} pixels")

            # Draw a line between the two points
            line, = ax.plot([pt1[0], pt2[0]], [pt1[1], pt2[1]], 'b--') # Blue dashed line
            lines.append(line)

            # Add text annotation for the distance on the plot
            mid_x = (pt1[0] + pt2[0]) / 2
            mid_y = (pt1[1] + pt2[1]) / 2
            dist_text = ax.text(mid_x, mid_y, f'Dist: {pixel_dist:.2f}', color='blue', fontsize=12, ha='center')
            texts.append(dist_text)

            # Update the main title with the distance
            ax.set_title(f"Distance: {pixel_dist:.2f} pixels")

            fig.canvas.draw()
            coords.clear()


# Connect the event listener
cid = fig.canvas.mpl_connect('button_press_event', onclick)
```
